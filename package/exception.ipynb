```python
try:
    num=int(input("Enter a number"))
    result=10/num
except ValueError:
    print("This is not a number")
except ZeroDivisionError:
    print("You can only enter denominator greater than zero")
except Exception as ex:
    print(ex)   


```

```python
try:
    age=int(input("enter your age"))
    b=50/age

except NameError:
    print('b is not intialized')
except ValueError:
    print("age can only be a number")
except ZeroDivisionError:
    print("can't divide by zero")
except Exception as ex:
    print(ex)

else:
    print(f"The value is {b}")

finally:
    print("I will be printed for sure")
```
