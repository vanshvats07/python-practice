```python
import seaborn as sns
```

```python
## Basic plotting with seaborn
tips=sns.load_dataset('tips')
print(tips)
```

```python
import matplotlib.pyplot as plt
sns.scatterplot(x='total_bill',y='tip',data=tips)
plt.title("Scatter plot of total bill vs tips")
plt.show()
```

```python
sns.lineplot(x='size',y='total_bill',data=tips,color='red')
plt.title("Line plot of total bill by size")

```

```python
## categorical plot
## bar plot

sns.barplot(x='day',y='total_bill',data=tips)
plt.title("Bar plot of total bill by day")
plt.show()
```

```python
## Box plot
sns.boxplot(x='day',y='total_bill',data=tips,color='purple')
```

```python
## Violin plot

sns.violinplot(x='day',y='total_bill',data=tips,color='red')
```
