```python
import seaborn as sns
```

```python
## Basic plotting with seaborn
tips=sns.load_dataset('tips')
print(tips)
```

```python
import matplotlib.pyplot as plt
sns.scatterplot(x='total_bill',y='tip',data=tips)
plt.title("Scatter plot of total bill vs tips")
plt.show()
```
