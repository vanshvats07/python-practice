```python
import pandas as pd
## Series
## 1d series array like object can hold any data type
series=pd.Series([1,2,3,4,5])
print(series)
print(type(series))

series.describe()

## create a series from dictionary
van={
    'name':'Vansh Vats',
    'class':'CSE_AI',
    'roll':18
}

print(type(van))

series=pd.Series(van)
print(type(series))


```

```python
data=[10,20,30]
index=['a','b','c']
pd.Series(data,index=index)
```

```python
## DataFrame create a DataFrame from a dictionary of list
data={
    'name':['Vansh','Daksh','Keshav'],
    'age':[18,16,17],
    'City':['Katewara','Katewara','Katewara']
}

df=pd.DataFrame(data)
print(df)
print(type(df))






```
