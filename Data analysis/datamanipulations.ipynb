```python
import pandas as pd
df=pd.read_csv('data.csv')
df.head(5)
df.describe()
df.dtypes
df.isnull().any(axis=1)
df.isnull().sum()
df_filled=df.fillna(0)
print(df_filled)

df['Value_fillna']=df['Value'].fillna(df['Value'].mean().astype(int))





```

```python
df.isnull().sum()
```

```python
##Remaining Columns
df=df.rename(columns={'Sale Date':'Sales Date'})
df.head()
```

```python
## change datatypes

df['Value_new']=df['Value'].astype(float)
df.head(5)
```

```python
df['New Value']=df['Value'].apply(lambda x:x*2)
df.head()
```

```python
grouped_mean=df.groupby('Product')['Value'].mean()
print(grouped_mean)
```

```python
grouped_sum=df.groupby(['Product','Region'])['Value'].sum()
print(grouped_sum)
```

```python
grouped_agg=df.groupby(['Region'])['Value'].agg(['mean','sum','count'])
print(grouped_agg)
```
