```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[1,4,9,16,25]
plt.plot(x,y)
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Basic Line Plot')
plt.show()

```

```python
x=[1,2,3,4,5]
y=[1,4,9,16,25]
plt.plot(x,y,color='red',linestyle='-')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.title('simple curve')
plt.show()

```

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[1,4,9,16,25]
plt.plot(x,y,color='red',linestyle='-.',marker='*')
plt.xlabel('X axis')
plt.grid(True)



```

```python
x=[1,2,3,4,5]
y1=[1,4,9,16,25]
y2=[1,2,3,4,5]
plt.figure(figsize=(9,5))
plt.subplot(1,2,1)
plt.plot(x,y1,color='green',linestyle='-',marker='.')
plt.title('Plot1')
plt.subplot(1,2,2)
plt.plot(x,y2,color='red',linestyle='-',marker='.')
plt.title('Plot2')
plt.subplot(2,2,1)
plt.plot(y1,x,color='green',linestyle='-',marker='.')

```
