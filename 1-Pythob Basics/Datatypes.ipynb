```python
ordered_set=set([1,2,3,4,5,6])
ordered_set
ordered_set.add(6)
ordered_set
```

```python
## discard method in set is that it will remove an item from the set if it is present in the set otherwise it will not throw any error.
vansh_set=set([1,2,3,4,5,6])
vansh_set.discard(6)
vansh_set.pop()

vansh_set
print(3 in vansh_set)
print(10 in vansh_set)
```

```python
set1={1,2,3,4,5}
set2={6,7,8,9,10}
union_set=set1.union(set2)
print(union_set)
```

```python
set1={1,2,3,4,5}
set2={2,4,5,6,7}
intersection_set=set1.intersection(set2)
print(intersection_set)
```
