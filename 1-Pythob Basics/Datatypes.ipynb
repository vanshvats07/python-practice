```python
employee={'name':'Vansh','age':16,'gender':'Male','city':'Delhi'}
print(employee['city'])

## get method of dictionary is used to access the value of a key in a dictionary. It takes the key as an argument and returns the corresponding value. If the key is not found, it returns None or a default value if specified.
print(employee.get('age')) 

##dictionaries are mutable, which means we can change their content without changing their identity. We can add new key-value pairs, update existing ones, or remove them.
employee["age"]=23
employee["address"]="Delhi"
print(employee)
keys=employee.keys()
print(keys)
values=employee.values()
print(values)
items=employee.items()
print(items)




```

```python
 ## shallow copy of a dictionary is a new dictionary that is created from an existing dictionary, but it only copies the references to the objects in the original dictionary. This means that if we modify a mutable object in the original dictionary, it will also affect the shallow copy. We can create a shallow copy of a dictionary using the copy() method or the dict() constructor.
employee_copy=employee
print(employee)
print(employee_copy)
employee_copy["age"]=25
print(employee_copy)
print(employee)
## deep copy of a dictionary is a new dictionary that is created from an existing dictionary, but it copies all the objects in the original dictionary recursively. This means that if we modify a mutable object in the original dictionary, it will not affect the deep copy. We can create a deep copy of a dictionary using the deepcopy() function from the copy module.
employee_deepcopy=employee.copy()
print(employee)
print(employee_deepcopy)
employee_deepcopy["name"]="Krish"
print(employee_deepcopy)
print(employee)
print(employee.keys())
print(employee.values())
print(employee.items())

```

```python
def is_strong(password):
    if(len(password)<8):
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
```

```python
def is_strong_password(password):
    if len(password)<8:
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
   

    return True

is_strong_password("Vansh@123")
```
